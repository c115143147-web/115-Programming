# 串列
可以儲存多個元素 整束浮點數字串 逗號隔開 前後用[ ]

```python
scores=[78,29,56,80,99]
print(scores)
```

# 索引值
從0開始

```python
print(scores[0])
```

# 串列中元素的個數
len(a) 可以取得a列有幾個元素

```python
print(len(scores))
print(scores[4])
print(scores[len(scores)-1])
print(scores[-1])
```

# 使用(append)新增元素到串列的尾端

```python
print(scores)
scores.append(69)
print(scores)
```

```python
print(scores)
scores[2]=70
print(scores)
```

```python
weights=[12]*6
print(weights)
```

# 走訪串列的三種方法

```python
print(scores)
for i in range(len(scores)):
    print(scores[i])
```

```python
for e in scores:
    print(e)
```

```python
for i in enumerate(scores):
    print(i,e)
```

# 計算費伯納西數列的第91個元素
如下:
1,1,2,3,5,8

```python
fib=[0]*91
print(fib)
fib[0]=1
fib[1]=1
print(fib)
fib[2]=fib[0]+fib[1]
fib[3]=fib[1]+fib[2]
```

```python
for i in range(2,91):
    fib[i]=fib[i-2]+fib[i-1]
print(fib[90])
```

# 99乘法表

```python
for i in range(1,6):
    for j in range(1,6):
        print(f'{i}*{j}={i*j}',end=',')
    print()
```

# 印出質數

```python
for i in range(2,273):
    if 273%i==0:
        print(f'{273}不是質數,因為被{i}除盡')
```

```python
for i in range(2,273):
    if 273%i==0:
        print(f'{273}不是質數')
        break
```

```python
n=17
isPrime=True
for i in range(2,n):
    if n%i==0:
        print(f'{n}不是質數')
        isPrime=False
        break
if isPrime==True:
    print(f'{n}是質數')
```

```python
import time
import pyautogui as pyg

# 關閉 PyAutoGUI 內建預設等待時間
pyg.PAUSE = 0

# 定義偵測區域 (left, top, width, height)
# 範圍: x: 670 -> 950 (寬 280), y: 950 -> 1200 (高 250)
REGION = (670, 950, 280, 250)

# 目標 RGB 顏色
TARGET_COLORS = {(83, 83, 83), (172, 172, 172)}

print("程式將在 3 秒後啟動，請切換至遊戲視窗...")
time.sleep(3)


while True:
    # 直接截取指定矩形區域
    im = pyg.screenshot(region=REGION)

    # 透過 getdata() 獲取所有像素點的 RGB tuple，並以 set 交集比對
    pixels = set(im.getdata())

    # 只要存在任一目標顏色即判定有障礙物
    if TARGET_COLORS & pixels:
        pyg.press('space')
        time.sleep(0.08)
```

```python
import time
import pyautogui as pyg

# 關閉 PyAutoGUI 內建延遲，確保按鍵最快響應
pyg.PAUSE = 0

# 固定 Y 軸範圍：900 到 1150
Y_START = 900
Y_END = 1100
HEIGHT = Y_END - Y_START  # 250 像素

# 目標比對顏色：經典深灰障礙物 (83, 83, 83) 與 夜間淺灰障礙物 (172, 172, 172)
TARGET_COLORS = {(83, 83, 83), (172, 172, 172)}

print("程式將在 3 秒後啟動，請點擊遊戲視窗並按下空白鍵讓恐龍起跑...")
time.sleep(3)

# 記錄遊戲開始運行的基準時間點
start_time = time.time()

try:
    while True:
        # 計算遊戲已進行的秒數
        elapsed = time.time() - start_time

        # 依時間切換 X 座標：前 20 秒為 800，20 秒後推進至 900
        current_x = 800 if elapsed < 20 else 900

        # 定義當前單一像素垂直線偵測範圍 (left, top, width, height)
        region = (current_x, Y_START, 1, HEIGHT)

        # 截取線段範圍並比對顏色
        im = pyg.screenshot(region=region)
        pixels = set(im.getdata())

        # 偵測到目標顏色即觸發跳躍
        if TARGET_COLORS & pixels:
            pyg.press('space')
            # 短暫冷卻，防止同一障礙物在空中重複觸發連擊
            #time.sleep(0.04)

except KeyboardInterrupt:
    print("\n[INFO] 使用者手動中斷，程式已安全結束。")
```

```python
import pyautogui as pyg
import time
from PIL import ImageGrab
#核心強制：關閉所有內建按鍵延遲
pyg.PAUSE = 0
#設定偵測範圍
DETECTION_X_START = 400  
DETECTION_X_END   = 950        
DETECTION_Y_START = 1000   
DETECTION_Y_END   = 1125         
        
#拉長最大蓄力時間
MAX_AIRTIME   = 0.22             # 最大大跳時間，足以飛越 4 個連續大仙人掌
MIN_COOLDOWN  = 0.01             # 近乎零的冷卻，落地一瞬間如果還有障礙物，立刻連跳
# =================================================================
   
print("【即將開始遊戲")
print("請在 2 秒內切換到遊戲畫面")
time.sleep(2)#等待2秒
 
last_jump_time = 0
is_jumping = False
while True:
    current_time = time.time()
    
    # 1. 抓取精準的動態防禦條
    bbox = (DETECTION_X_START, DETECTION_Y_START, DETECTION_X_END, DETECTION_Y_END)
    im = ImageGrab.grab(bbox=bbox)
    
    # 2. 自動抓取當下左上角的像素作為「即時背景」
    bg_pixel = im.getpixel((0, 0))
    colors = im.getcolors()
    
    has_obstacle = False
    if colors:
        for count, color in colors:
            # 排除背景色，且必須有超過 2 個像素點才算障礙物（抗雜訊）
            if color != bg_pixel and count > 2:
                try:
                    # 日夜漸層色差判定（維持 > 3 的超高靈敏度）
                    rdiff = abs(color[0] - bg_pixel[0])
                    gdiff = abs(color[1] - bg_pixel[1])
                    bdiff = abs(color[2] - bg_pixel[2])
                    
                    if rdiff > 3 or gdiff > 3 or bdiff > 3:
                        has_obstacle = True
                        break
                except (TypeError, IndexError):
                    if abs(color - bg_pixel) > 3:
                        has_obstacle = True
                        break
                
    # 3. 狀態機控制：達到最大大跳時間，強制放開 Space 準備落地
    if is_jumping and (current_time - last_jump_time > MAX_AIRTIME):
        pyg.keyUp('space') 
        is_jumping = False
        
    # 4. 觸發防禦機制
    if has_obstacle:
        # 如果還在跳躍中，但前方依然有障礙物（代表是 4 連排超大障礙），就繼續按著 Space 蓄力
        if is_jumping:
            pyg.keyDown('space')
        # 如果是落地狀態或冷卻結束，立刻發動新的一跳
        elif current_time - last_jump_time > MIN_COOLDOWN:
            pyg.keyUp('down')   # 確保沒卡下蹲
            pyg.keyDown('space') # 長按蓄力起跳
            last_jump_time = time.time()
            is_jumping = True
    else:
        # 如果偵測區間內已經沒有障礙物了，提早釋放 Space（視為短跳），避免滯空太久卡到下一個障礙
        if is_jumping and (current_time - last_jump_time > 0.08):
            pyg.keyUp('space')
            is_jumping = False
```
